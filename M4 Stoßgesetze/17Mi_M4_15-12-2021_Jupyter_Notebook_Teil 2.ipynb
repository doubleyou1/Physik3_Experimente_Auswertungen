## Aufgabe 2

### Gemessen:
    - Auslenkung der gr. Kugel, bei abrollender Kugel vom Abstand s. 5x pro Abstand
    - Höhe H, $h_0$, L und l (vgl. Skizze)
### Formel:
Auslenkung der großen Kugel:
$$
a_{gr} ' = \frac{2 m_{roll}}{m_{roll} + m_{gr}} \sqrt{\epsilon \,2 \,l \,h} \Leftrightarrow a_{gr} ' = \frac{2 m_{roll}}{m_{roll} + m_{gr}} \sqrt{\varepsilon \,2 \,l} \sqrt{h}
$$
mit der Steigung k 
$$
k_3 = \frac{2 m_{roll}}{m_{roll} + m_{gr}} \sqrt{\epsilon \,2 \,l} \Leftrightarrow \varepsilon = \left(\frac{k_3(m_{roll} + m_{gr})}{2m_{roll}}\right)^2 \frac{1}{2\,l}
$$

Unsicherheit:
$$
\frac{\partial \varepsilon}{\partial k_3} = \frac{k_3}{l}\left( \frac{m_{roll} + m_{gr}}{2 m_{roll}}\right)^2
$$

$$
\frac{\partial \varepsilon}{\partial m_{roll}} = - \frac{m_{gr} k_3^2(m_{roll} + m_{gr})}{4\,l\,m_{roll}^3}
$$

$$
\frac{\partial \varepsilon}{\partial m_{gr}} = \frac{k_3^2(m_{roll} + m_{gr})}{4\,l\,m_{roll}^2}
$$

$$
\frac{\partial \varepsilon}{\partial l} = - \frac{k_3^2(m_{roll} + m_{gr})^2}{8 m_{roll}^2 l^2}
$$
Insgesamt: 
$$
u_{\varepsilon} = \sqrt{\left(\frac{\partial \varepsilon}{\partial k_3} \cdot u_{k_3} \right)^2 + \left(\frac{\partial \varepsilon}{\partial m_{roll}} \cdot u_{m_{roll}} \right)^2 + \left(\frac{\partial \varepsilon}{\partial m_{gr}} \cdot u_{m_{gr}} \right)^2 + \left(\frac{\partial \varepsilon}{\partial l} \cdot u_l \right)^2}
$$


Herleitung:
Aus Skizze ref{bla} nimmt man die Beziehung 
$$
sin(\alpha) = \frac{h- h_0}{S}, \qquad tan(\alpha) = \frac{H}{L}
$$
Daraus ergibt sich:
$$
h = h_0 - sin(\alpha)\cdot S
$$
Allgemein lässt sich der $tan(\alpha)$ umschreiben lassen zu:
$$
tan(\alpha) = \frac{sin(\alpha)}{cos(\alpha)} 
$$
mit $cos(\alpha) = \sqrt{1 - sin^2(\alpha)}$ folgt:
$$
tan(\alpha) = \frac{sin(\alpha)}{\sqrt{1 - sin^2(\alpha)}}
$$
nach auflösen nach $sin(\alpha)$ ergibt sich folgendes:
$$
sin(\alpha) = \frac{tan(\alpha)}{\sqrt{1 + tan^2(\alpha)}}
$$
Ingesamt lässt sich dann h schreiben als:
$$
h = h_0 - \frac{tan(\alpha)}{\sqrt{1 + tan^2(\alpha)}}\cdot S
$$
Mit der anfänglich erwähnten Beziehung $tan(\alpha) = \frac{H}{L}$ folgt:
$$
h = h_0 - \frac{\frac{H}{L}\cdot S}{\sqrt{1 + \frac{H^2}{L^2}}}
$$
Wird nun ein ein $H^2$ und $L^2$ aus der Wurzel herausgezogen ergibt sich abschließend:
$$
h = h_0 - \frac{S}{\sqrt{1 + \frac{L^2}{H^2}}}
$$
Für den plott:
$$
h \rightarrow \sqrt{h} = h_w
$$
ebenso:
$$
u_{h_w} = \frac{u_h}{2\sqrt{h}} 
$$

Unsicherheit:
$$
\frac{\partial h}{\partial h_0} = 1
$$

$$
\frac{\partial h}{\partial s} = - \frac{1}{\sqrt{1 + \frac{L^2}{H^2}}}
$$

$$
\frac{\partial h}{\partial L} =  \frac{S}{2 \left(1 + \frac{L^2}{H^2}\right)^{\frac{3}{2}}} \frac{2 L}{H^2} =  \frac{S}{ \left(1 + \frac{L^2}{H^2}\right)^{\frac{3}{2}}} \frac{L}{H^2}
$$

$$
\frac{\partial h}{\partial H} =  \frac{S}{2 \left(1 + \frac{L^2}{H^2}\right)^{\frac{3}{2}}} \frac{- 2 L^2}{H^3} = - \frac{S}{ \left(1 + \frac{L^2}{H^2}\right)^{\frac{3}{2}}} \frac{L^2}{H^3}
$$

$$
u_h = \sqrt{\left(\frac{\partial h}{\partial h_0} \cdot u_{h_0}\right)^2 + \left(\frac{\partial h}{\partial s} \cdot u_s \right)^2 + \left(\frac{\partial h}{\partial L}\cdot u_L \right)^2 + \left( \frac{\partial h}{\partial H} \cdot u_H\right)^2}
$$
